# RetailIQ — Data Understanding

## Objective

The objective of this notebook is to understand the Olist Brazilian E-Commerce dataset before performing data cleaning or transformation.

We will:

- Inspect the available datasets
- Understand the structure of each table
- Examine rows and columns
- Check data types
- Identify missing values
- Check duplicate records
- Understand relationships between tables
- Examine date ranges
- Identify the datasets and fields relevant to RetailIQ

> The raw dataset will not be modified in this notebook.

## Cell 1 — Import libraries

```python
# Import pandas for loading, cleaning, and analyzing tabular data
import pandas as pd

# Import NumPy for numerical operations
import numpy as np

# Import Matplotlib for creating visualizations
import matplotlib.pyplot as plt

# Import Seaborn for statistical and exploratory visualizations
import seaborn as sns

# Import Path to work with file and folder paths in a platform-independent way
from pathlib import Path

# Display all columns when viewing DataFrames
pd.set_option("display.max_columns", None)

# Display up to 100 rows when viewing DataFrames
pd.set_option("display.max_rows", 100)

# Confirm that the required libraries have been imported successfully
print("Libraries loaded successfully.")
```

## Cell 2 — Set the data path

```python
# Import Path so we can work with project folders reliably across operating systems
from pathlib import Path

# Define the location of the raw Olist dataset relative to this notebook
DATA_PATH = Path("../data/raw")

# Check that the expected data folder exists
print("Data folder exists:", DATA_PATH.exists())

# Display the path we are using
print("Data path:", DATA_PATH)
```

## Cell 3 — Find the CSV files

```python
# Find all CSV files inside the raw data folder
csv_files = sorted(DATA_PATH.glob("*.csv"))

# Count how many CSV files were found
print(f"Number of CSV files found: {len(csv_files)}")

# Display the name of every CSV file
for file in csv_files:
    print(file.name)
```

## Cell 4 — Load the datasets

```python
# Create an empty dictionary to store all Olist DataFrames
datasets = {}

# Loop through each CSV file
for file in csv_files:
    
    # Use the filename without ".csv" as the dictionary key
    dataset_name = file.stem
    
    # Read the CSV file into a pandas DataFrame
    datasets[dataset_name] = pd.read_csv(file)

# Confirm that all datasets have been loaded
print(f"Loaded {len(datasets)} datasets successfully.")
```

```python
# Display the number of rows and columns for every dataset
for name, df in datasets.items():
    print(f"{name}: {df.shape}")
```

## Cell 6 -- Column names

```python
# Display the column names for every dataset
# This helps us understand what information each table contains

for name, df in datasets.items():
    print(f"\n{'=' * 70}")
    print(f"DATASET: {name}")
    print(f"{'=' * 70}")
    
    # Display all column names
    print("Columns:")
    for column in df.columns:
        print(f"  - {column}")
```

## Cell 7 -- Data types and missing values

```python
# Create a compact data-quality summary for every dataset
# This lets us review data types and missing values without producing huge output

for name, df in datasets.items():
    print(f"\n{'=' * 70}")
    print(f"DATASET: {name}")
    print(f"{'=' * 70}")
    
    # Create a summary containing column names, data types, and missing-value counts
    summary = pd.DataFrame({
        "Data Type": df.dtypes,
        "Missing Values": df.isnull().sum(),
        "Missing %": (df.isnull().sum() / len(df) * 100).round(2)
    })
    
    # Display the summary
    display(summary)
```

## Cell 8 - Preview the data

```python
# Display the first three rows of every dataset
# This helps us see what the actual records look like

for name, df in datasets.items():
    print(f"\n{'=' * 70}")
    print(f"DATASET: {name}")
    print(f"{'=' * 70}")
    
    # Display the first three records
    display(df.head(3))
```

```python
# Check the number of duplicate rows in each dataset.
# Duplicate records can affect analysis, so we identify them before making any changes.

duplicate_summary = []

for name, df in datasets.items():
    duplicate_summary.append({
        "Dataset": name,
        "Total Rows": len(df),
        "Duplicate Rows": df.duplicated().sum(),
        "Unique Rows": df.drop_duplicates().shape[0]
    })

# Convert the results into a DataFrame for easy comparison
duplicate_summary = pd.DataFrame(duplicate_summary)

# Display the duplicate summary
display(duplicate_summary)
```

```python
# Check the number of unique values in important ID columns.
# This helps us understand which columns may act as identifiers or keys.

id_columns = {
    "olist_customers_dataset": ["customer_id", "customer_unique_id"],
    "olist_order_items_dataset": ["order_id", "order_item_id", "product_id", "seller_id"],
    "olist_order_payments_dataset": ["order_id"],
    "olist_order_reviews_dataset": ["review_id", "order_id"],
    "olist_orders_dataset": ["order_id", "customer_id"],
    "olist_products_dataset": ["product_id"],
    "olist_sellers_dataset": ["seller_id"]
}

for dataset_name, columns in id_columns.items():
    df = datasets[dataset_name]
    
    print(f"\n{'=' * 70}")
    print(f"DATASET: {dataset_name}")
    print(f"{'=' * 70}")
    
    for column in columns:
        print(f"{column}: {df[column].nunique():,} unique values")
```

## Verify the keys

```python
# Check whether important identifier combinations are unique.
# This helps us determine which columns or combinations of columns
# can act as keys when we connect the different datasets.

key_checks = {
    "Customers - customer_id": (
        datasets["olist_customers_dataset"],
        ["customer_id"]
    ),
    
    "Customers - customer_unique_id": (
        datasets["olist_customers_dataset"],
        ["customer_unique_id"]
    ),
    
    "Orders - order_id": (
        datasets["olist_orders_dataset"],
        ["order_id"]
    ),
    
    "Order Items - order_id + order_item_id": (
        datasets["olist_order_items_dataset"],
        ["order_id", "order_item_id"]
    ),
    
    "Products - product_id": (
        datasets["olist_products_dataset"],
        ["product_id"]
    ),
    
    "Sellers - seller_id": (
        datasets["olist_sellers_dataset"],
        ["seller_id"]
    )
}

# Check each candidate key for duplicate combinations
for name, (df, columns) in key_checks.items():
    
    # Count how many rows share the same values for the selected key
    duplicate_keys = df.duplicated(subset=columns).sum()
    
    # Count the number of unique key combinations
    unique_keys = df[columns].drop_duplicates().shape[0]
    
    print(f"\n{name}")
    print(f"  Columns: {columns}")
    print(f"  Unique key combinations: {unique_keys:,}")
    print(f"  Duplicate key rows: {duplicate_keys:,}")
```

## Map the relationships

```python
# Check whether foreign-key values in child tables exist
# in the corresponding parent tables.
# This helps us identify valid relationships and possible unmatched records.

relationship_checks = [
    {
        "relationship": "Orders → Customers",
        "child_df": datasets["olist_orders_dataset"],
        "child_column": "customer_id",
        "parent_df": datasets["olist_customers_dataset"],
        "parent_column": "customer_id"
    },
    {
        "relationship": "Order Items → Orders",
        "child_df": datasets["olist_order_items_dataset"],
        "child_column": "order_id",
        "parent_df": datasets["olist_orders_dataset"],
        "parent_column": "order_id"
    },
    {
        "relationship": "Order Items → Products",
        "child_df": datasets["olist_order_items_dataset"],
        "child_column": "product_id",
        "parent_df": datasets["olist_products_dataset"],
        "parent_column": "product_id"
    },
    {
        "relationship": "Order Items → Sellers",
        "child_df": datasets["olist_order_items_dataset"],
        "child_column": "seller_id",
        "parent_df": datasets["olist_sellers_dataset"],
        "parent_column": "seller_id"
    },
    {
        "relationship": "Payments → Orders",
        "child_df": datasets["olist_order_payments_dataset"],
        "child_column": "order_id",
        "parent_df": datasets["olist_orders_dataset"],
        "parent_column": "order_id"
    },
    {
        "relationship": "Reviews → Orders",
        "child_df": datasets["olist_order_reviews_dataset"],
        "child_column": "order_id",
        "parent_df": datasets["olist_orders_dataset"],
        "parent_column": "order_id"
    }
]

# Test each relationship
for relationship in relationship_checks:
    
    # Get the unique foreign-key values from the child table
    child_keys = set(
        relationship["child_df"][relationship["child_column"]].dropna().unique()
    )
    
    # Get the valid primary-key values from the parent table
    parent_keys = set(
        relationship["parent_df"][relationship["parent_column"]].dropna().unique()
    )
    
    # Find child keys that do not exist in the parent table
    unmatched_keys = child_keys - parent_keys
    
    print(f"\n{relationship['relationship']}")
    print(f"  Child unique keys: {len(child_keys):,}")
    print(f"  Unmatched keys: {len(unmatched_keys):,}")
```

```python
# Check whether product categories in the products table
# have corresponding English translations.
# This will help us understand whether the translation table
# can be safely joined to the product data.

products = datasets["olist_products_dataset"]
translations = datasets["product_category_name_translation"]

# Get the unique product categories from the products table
product_categories = set(
    products["product_category_name"].dropna().unique()
)

# Get the categories available in the translation table
translated_categories = set(
    translations["product_category_name"].dropna().unique()
)

# Find product categories that do not have a translation
untranslated_categories = product_categories - translated_categories

print(f"Unique product categories: {len(product_categories):,}")
print(f"Translated categories: {len(translated_categories):,}")
print(f"Categories without translation: {len(untranslated_categories):,}")

# Display the categories that do not have a translation
if untranslated_categories:
    print("\nCategories without translation:")
    for category in sorted(untranslated_categories):
        print(f"  - {category}")
```

```python
# Analyze the distribution of order statuses.
# This helps us understand the different stages of the order lifecycle
# before deciding which orders should be included in sales analysis.

orders = datasets["olist_orders_dataset"]

# Count the number of orders for each status
order_status_counts = orders["order_status"].value_counts()

# Display the order status counts
display(order_status_counts)
```

```python
# Calculate the percentage of orders represented by each status.
# This gives us a better understanding of the relative importance
# of each stage in the order lifecycle.

order_status_percentage = (
    orders["order_status"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
    .rename("Percentage")
)

# Display the percentage distribution
display(order_status_percentage)
```

```python
# Convert the order purchase timestamp from text to datetime.
# We create a temporary datetime column for analysis without changing
# the original raw dataset stored in our data/raw folder.

orders["purchase_datetime"] = pd.to_datetime(
    orders["order_purchase_timestamp"]
)

# Display the earliest and latest order purchase dates
print("Earliest order:", orders["purchase_datetime"].min())
print("Latest order:", orders["purchase_datetime"].max())
```

```python
# Calculate the overall time span covered by the order data.
# This tells us how much historical data is available for analysis
# and eventually for sales forecasting.

date_range = (
    orders["purchase_datetime"].max()
    - orders["purchase_datetime"].min()
)

print(f"Order history: {date_range.days} days")
print(f"Approximately {date_range.days / 365.25:.2f} years of data")
```

```python
# Convert the important order timestamp columns from text to datetime.
# We use errors="coerce" so invalid or missing values become NaT
# instead of causing the analysis to fail.

date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:
    orders[column] = pd.to_datetime(
        orders[column],
        errors="coerce"
    )

# Display the earliest and latest date available for each timestamp column.
# This helps us understand the coverage of each stage of the order lifecycle.

date_summary = []

for column in date_columns:
    date_summary.append({
        "Column": column,
        "Earliest Date": orders[column].min(),
        "Latest Date": orders[column].max(),
        "Missing Values": orders[column].isna().sum()
    })

# Convert the results into a DataFrame for easy comparison.
date_summary = pd.DataFrame(date_summary)

display(date_summary)
```

## Order-item financial summary

```python
# Analyze the financial fields in the order items table.
# This helps us understand product prices and shipping costs at the item level.

order_items = datasets["olist_order_items_dataset"]

# Calculate basic statistics for price and freight value
financial_summary = order_items[["price", "freight_value"]].describe().T

# Add the total amount for each financial field
financial_summary["Total"] = order_items[["price", "freight_value"]].sum()

# Display the financial summary
display(financial_summary)
```

## Understand how many items orders contain

```python
# Count the number of items associated with each order.
# An order can contain multiple products, so this helps us understand
# the relationship between orders and order items.

items_per_order = (
    order_items
    .groupby("order_id")
    .size()
)

# Display summary statistics for items per order
display(items_per_order.describe())
```

## Check payment structure

```python
# Analyze the payment table to understand how payments are distributed
# across orders.

payments = datasets["olist_order_payments_dataset"]

# Count how many payment records are associated with each order
payments_per_order = (
    payments
    .groupby("order_id")
    .size()
)

# Display summary statistics
display(payments_per_order.describe())
```

## Understand payment values

```python
# Calculate basic statistics for payment values.
# This helps us understand the monetary amounts recorded in the payment table.

payment_summary = payments["payment_value"].describe()

# Display the payment statistics
display(payment_summary)
```

```python
# Aggregate product prices and freight costs at the order level.
# This prevents individual order-item rows from being treated as separate orders.

order_item_totals = (
    order_items
    .groupby("order_id")
    .agg(
        item_price=("price", "sum"),
        freight_value=("freight_value", "sum")
    )
    .reset_index()
)

# Calculate the total item-level amount including freight for each order.
order_item_totals["item_total"] = (
    order_item_totals["item_price"]
    + order_item_totals["freight_value"]
)

# Aggregate all payment records belonging to the same order.
# An order can have multiple payment records, so they must be summed first.

payment_totals = (
    payments
    .groupby("order_id")
    .agg(
        payment_value=("payment_value", "sum")
    )
    .reset_index()
)

# Combine the order-level item totals with order-level payment totals.
financial_comparison = order_item_totals.merge(
    payment_totals,
    on="order_id",
    how="outer"
)

# Display summary statistics for the two monetary measures.
display(
    financial_comparison[
        ["item_price", "freight_value", "item_total", "payment_value"]
    ].describe()
)
```

```python
# Calculate the overall totals for each financial measure.
# This lets us compare the item-level amount with the recorded payment amount.

financial_totals = pd.DataFrame({
    "Metric": [
        "Product Price",
        "Freight",
        "Product Price + Freight",
        "Payment Value"
    ],
    "Total": [
        financial_comparison["item_price"].sum(),
        financial_comparison["freight_value"].sum(),
        financial_comparison["item_total"].sum(),
        financial_comparison["payment_value"].sum()
    ]
})

# Display the overall financial totals.
display(financial_totals)
```

## Payment type distribution

```python
# Analyze the different payment methods used by customers.
# This helps us understand the structure of the payment table.

payment_type_summary = (
    payments["payment_type"]
    .value_counts()
    .rename_axis("Payment Type")
    .reset_index(name="Number of Payments")
)

# Display the number of payment records for each payment type
display(payment_type_summary)
```

## Payment value by type

```python
# Calculate the total and average payment value for each payment method.
# This helps us understand how monetary value is distributed across
# different payment types.

payment_value_by_type = (
    payments
    .groupby("payment_type")["payment_value"]
    .agg(["count", "sum", "mean"])
    .reset_index()
)

# Rename the columns to make the results easier to understand
payment_value_by_type.columns = [
    "Payment Type",
    "Number of Payments",
    "Total Payment Value",
    "Average Payment Value"
]

# Display the payment value breakdown
display(payment_value_by_type)
```
